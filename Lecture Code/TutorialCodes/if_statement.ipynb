```python
x = 5
y = 5

if x == y:
    print("x equals y")
else:
    print("x doesn't equal y")
```

```python
x = 10
y = 5

if x != y:
    print("x doesn't equal y")
else:
    print("x equals y")
```

```python
x = 10
y = 5

if x == y:
    print("x equals y")
elif x < y:
    print("x is less than y")
else:
    print("x is greater than y")
```
